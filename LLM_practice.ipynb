```python
import pandas as pd

properties = pd.DataFrame({
    "location": ["Marina", "JVC", "Downtown"],
    "beds": [2, 1, 3],
    "price": [1200000, 650000, 3200000],
})

for i, row in properties.iterrows():
    question = f"Write a one-line listing for a {row['beds']}-bed apartment in {row['location']} priced at AED {row['price']:,}"

    response = client.models.generate_content(
        model="gemini-flash-lite-latest",
        contents=question,
    )

    print(f"{row['location']}: {response.text}")
    print()
```
